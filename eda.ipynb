```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("models_full.csv")

# 1. Dataset overview
print("Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)

# 2. Missing values
missing = df.isna().sum().sort_values(ascending=False)
print("\nMissing Values:")
print(missing)

# 3. Price distribution
priced = df.dropna(subset=[
    "latest_input_price_per_1m",
    "latest_output_price_per_1m"
]).copy()

plt.figure(figsize=(8, 4))
plt.hist(priced["latest_input_price_per_1m"], bins=15)
plt.xlabel("Input Price (USD per 1M tokens)")
plt.ylabel("Number of Models")
plt.title("Distribution of Input API Prices")
plt.show()

# 4. Input vs output pricing
plt.figure(figsize=(8, 4))
plt.scatter(
    priced["latest_input_price_per_1m"],
    priced["latest_output_price_per_1m"]
)
plt.xlabel("Input Price (USD per 1M tokens)")
plt.ylabel("Output Price (USD per 1M tokens)")
plt.title("Input vs Output API Pricing")
plt.show()

# 5. Model coverage by provider
provider_counts = df["provider_name"].value_counts()

plt.figure(figsize=(9, 4))
provider_counts.plot(kind="bar")
plt.xlabel("Provider")
plt.ylabel("Number of Models")
plt.title("Models by Provider")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 6. Key data completeness
key_cols = [
    "context_window",
    "supports_vision",
    "supports_tool_use",
    "latest_input_price_per_1m",
    "latest_output_price_per_1m"
]

print("\nKey Field Coverage:")
print(df[key_cols].notna().sum())
```
